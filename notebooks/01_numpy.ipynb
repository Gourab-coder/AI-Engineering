```python
## 1. creating array
```

```python
list1 = [1, 2, 3]
print(list1)
```

```python
arr_list1 = np.array(list1)
print(arr_list1)
```

```python
print(type(arr_list1))
print(arr_list1.dtype)
```

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
arr_list12 = np.array([list1, list2])
print(arr_list12)
```

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = [7, 8, 9]
arr_list123 = np.array([[list1, list2, list3]])
print(arr_list123)
```

```python
print(f"Dimention: arr_list1: {arr_list1.ndim}, arr_list12: {arr_list12.ndim}, arr_list123: {arr_list123.ndim}")
```

```python
## numpy array attributes
```

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
print(arr1)
```

```python
print("shape", arr1.shape)
print("size", arr1.size)
print("dtype", arr1.dtype)
print("ndim", arr1.ndim)
```

```python
## array initialization methods
```

```python
# zero array
zero_arr = np.zeros((2, 3))
print(zero_arr)
```

```python
# one array
one_arr = np.ones((1, 3))
print(one_arr)
```

```python
# full array
full_arr = np.full((3, 2), 7)
print(full_arr)
```

```python
# identity array
id_arr = np.eye(3)
print(id_arr)
```

```python
# empty array
empty_arr = np.empty(2)
print(empty_arr)
```

```python
# evenly space array
evenly_arr = np.arange(1, 10, 3)
print(evenly_arr)
```

```python
# specific number of eqaully (spaced) values between a range
print(np.linspace(1, 10, 4))
```

```python
# random values array - float
r_arr1 = np.random.rand(3, 2)
print(r_arr1)
```

```python
# random values array - int
r_arr2 = np.random.randint(1, 10, (3, 3))
print(r_arr2)
```
